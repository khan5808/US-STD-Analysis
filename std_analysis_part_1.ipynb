```python
#Morbidity EDA
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Assuming std_morbidity dataframe already exists
# Set the style for better visualization
plt.style.use('ggplot')

# Create a figure with 3 subplots
fig, axes = plt.subplots(3, 1, figsize=(14, 18))

# 1. Disease vs Disease Code
# Group by disease and get the most common disease code for each disease
disease_code_mapping = std_morbidity.groupby('Disease')['Disease Code'].agg(lambda x: x.value_counts().index[0])
sns.barplot(x=disease_code_mapping.index, y=disease_code_mapping.values, ax=axes[0])
axes[0].set_title('Disease vs Most Common Disease Code', fontsize=16)
axes[0].set_xlabel('Disease', fontsize=12)
axes[0].set_ylabel('Disease Code', fontsize=12)
axes[0].tick_params(axis='x', rotation=45)

# 2. Disease vs Count and Population
# Group by disease and calculate mean count and population
disease_stats = std_morbidity.groupby('Disease').agg({'Count': 'mean', 'Population': 'mean'}).reset_index()

# Create a bar plot with two bars for each disease
x = np.arange(len(disease_stats))
width = 0.35

axes[1].bar(x - width/2, disease_stats['Count'], width, label='Count')
axes[1].bar(x + width/2, disease_stats['Population'], width, label='Population')
axes[1].set_title('Disease vs Count and Population (Mean Values)', fontsize=16)
axes[1].set_xlabel('Disease', fontsize=12)
axes[1].set_ylabel('Value', fontsize=12)
axes[1].set_xticks(x)
axes[1].set_xticklabels(disease_stats['Disease'], rotation=45)
axes[1].legend()

# Add a second y-axis for population if the scales are very different
if disease_stats['Population'].mean() > disease_stats['Count'].mean() * 10:
    ax2 = axes[1].twinx()
    ax2.bar(x + width/2, disease_stats['Population'], width, color='orange', alpha=0.7, label='Population')
    ax2.set_ylabel('Population', fontsize=12)
    ax2.legend(loc='upper right')
    axes[1].legend(loc='upper left')

# 3. Disease vs Rate
# Group by disease and calculate mean rate
disease_rate = std_morbidity.groupby('Disease')['Rate'].mean().reset_index()
sns.barplot(x='Disease', y='Rate', data=disease_rate, ax=axes[2])
axes[2].set_title('Disease vs Rate (Mean Values)', fontsize=16)
axes[2].set_xlabel('Disease', fontsize=12)
axes[2].set_ylabel('Rate', fontsize=12)
axes[2].tick_params(axis='x', rotation=45)

# Add data labels to the rate chart
for i, v in enumerate(disease_rate['Rate']):
    axes[2].text(i, v + 0.01, f'{v:.2f}', ha='center')

plt.tight_layout()
plt.show()

# Optional: Create a heatmap to show the relationship between disease and disease code
plt.figure(figsize=(12, 8))
disease_code_pivot = pd.crosstab(std_morbidity['Disease'], std_morbidity['Disease Code'])
sns.heatmap(disease_code_pivot, cmap='YlGnBu', annot=True, fmt='d')
plt.title('Heatmap of Disease vs Disease Code Frequency', fontsize=16)
plt.xlabel('Disease Code', fontsize=12)
plt.ylabel('Disease', fontsize=12)
plt.tight_layout()
plt.show()
```

```python
#STI Cases EDA
years_cols = ['2020', '2021', '2022', '2023', '2024']

# Create the figure
plt.figure(figsize=(18, 12))

# Part 1: Bar chart for years 2020-2024 by disease with sex as hue
plt.subplot(2, 1, 1)

# Filter data to include only the rows we want to visualize
plot_data = sti_cases_cdc[sti_cases_cdc['Sex'].isin(['Men', 'Women'])]

# Get unique diseases for plotting
diseases = plot_data['Disease'].unique()
sex_categories = ['Men', 'Women']

# Set up the positions for grouped bars
bar_width = 0.15
disease_positions = np.arange(len(years_cols))
n_diseases = len(diseases)

# Create a custom colormap for diseases and sex
disease_colors = sns.color_palette("husl", n_diseases)
sex_colors = {'Men': 'steelblue', 'Women': 'salmon'}

# Plot each disease by sex
for i, disease in enumerate(diseases):
    for j, sex in enumerate(sex_categories):
        data_row = plot_data[(plot_data['Disease'] == disease) & (plot_data['Sex'] == sex)]
        if not data_row.empty:
            # Calculate position: group by disease, then by sex within disease
            position = disease_positions + (i - n_diseases/2 + j*0.5) * bar_width
            
            # Get color for sex
            color = sex_colors[sex]
            
            plt.bar(position, 
                    data_row[years_cols].values[0], 
                    width=bar_width, 
                    label=f"{disease} - {sex}", 
                    color=color,
                    alpha=0.7 if sex == 'Women' else 0.9)

# Format y-axis to show actual values instead of scientific notation
ax1 = plt.gca()
ax1.yaxis.set_major_formatter(ticker.StrMethodFormatter('{x:,.0f}'))  # No decimal places

plt.xticks(disease_positions, years_cols)
plt.title('STI Cases by Year, Disease, and Sex', fontsize=16)
plt.ylabel('Number of Cases', fontsize=12)
plt.legend(loc='upper left', bbox_to_anchor=(1, 1))
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Part 2: Percent changes visualization with sex as hue
plt.subplot(2, 1, 2)

# Create a more organized horizontal bar chart for percent changes
diseases_unique = plot_data['Disease'].unique()

# Calculate positions for the bars
y_positions = []
y_labels = []
colors_5yr = []
colors_1yr = []

gap = 0.5  # Gap between disease groups
current_pos = 0

for disease in diseases_unique:
    disease_data = plot_data[plot_data['Disease'] == disease]
    
    for sex in sex_categories:
        sex_data = disease_data[disease_data['Sex'] == sex]
        if not sex_data.empty:
            y_positions.append(current_pos)
            y_labels.append(f"{disease} - {sex}")
            colors_5yr.append(sex_colors[sex])
            colors_1yr.append(sex_colors[sex])
            current_pos += 1
    
    current_pos += gap  # Add gap between disease groups

# Plot 5-year and 1-year changes
for i, (disease, sex) in enumerate(zip(plot_data['Disease'], plot_data['Sex'])):
    row = plot_data[(plot_data['Disease'] == disease) & (plot_data['Sex'] == sex)]
    if not row.empty and i < len(y_positions):
        pos = y_positions[i]
        plt.barh(pos - 0.2, row['Percent_Change_5yr'].values[0], height=0.4, 
                color=colors_5yr[i], alpha=0.7, label='_5-Year Change')
        plt.barh(pos + 0.2, row['Percent_Change_1yr'].values[0], height=0.4, 
                color=colors_1yr[i], alpha=0.9, label='_1-Year Change')

# Format x-axis to show percentages with % symbol
ax2 = plt.gca()
ax2.xaxis.set_major_formatter(ticker.PercentFormatter(xmax=100))

# Add custom legend for 5-year and 1-year changes
handles = [
    plt.Rectangle((0,0),1,1, color='gray', alpha=0.7),
    plt.Rectangle((0,0),1,1, color='gray', alpha=0.9)
]
labels = ['5-Year Change', '1-Year Change']

# Add custom legend for sex categories
for sex, color in sex_colors.items():
    handles.append(plt.Rectangle((0,0),1,1, color=color))
    labels.append(sex)

plt.legend(handles, labels, loc='lower right')

plt.yticks(y_positions, y_labels)
plt.xlabel('Percent Change', fontsize=12)
plt.title('Percent Change in STI Cases by Disease and Sex', fontsize=16)
plt.axvline(x=0, color='black', linestyle='-', alpha=0.3)
plt.grid(axis='x', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```
